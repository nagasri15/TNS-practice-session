```python
#function without parameters
def greet():
  print("HI")
greet()
```

```python
#function with parameters
def greet(name):
  print("hello", name)
greet("sri")
greet("tara")
```

```python
#function with return val
def add(a,b):
  return a+b
result = add(15,90)
print(result)
```

```python
#function with return val
def square(num):
  return num*num
result = square(10)
print(result)
```

```python
#Lambda function
add = lambda a,b : a+b
print(add(45, 90))
```

```python
#Lambda function for printing name
N = lambda name : name
print(N("sita"))
```

```python
N = lambda num : num*num
print(N(5))
```

```python
#factorial of n
def factorial(n):
  if n==0 or n==1:
    return 1
  else:
    return n*factorial(n-1)
print(factorial(5))
```

```python
#class and obj
class student:
  name = "sri"
s = student()
print(s.name)
```

```python
class employees:
  emp_name = "shiva"
  emp_id = 101
  emp_age = 51
e = employees()
print(e.emp_age , e.emp_name)
```

```python
class Rectangle:
  length = 10
  breadth = 20
  def area(self):
    print("area of rec: ",self.length*self.breadth)
r = Rectangle()
r.area()
```

```python

#constructor
class student:
  def __init__(self,name, age, gender):
    self.name = name
    self.age = age
    self.gender = gender
s1 = student("sri", 21, "female")
print("name: ", s1.name, "age: ",s1.age)
```
