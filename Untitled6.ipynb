```python
import pandas as pd

# Load the dataset
df = pd.read_csv('/content/netflix_titles.csv')
print(df.shape)
print(df.info())
print(df.head())
df
```

```python
import matplotlib.pyplot as plt
import pandas as pd
content_type_counts = df['type'].value_counts()
content_type_counts.plot(kind='bar')
plt.xlabel('Content Type')
plt.ylabel('Count')
plt.title('Distribution of Content Types')
plt.show()
#country contribution using bar
content_type_counts = df['country'].value_counts().head(10)
content_type_counts.plot(kind='bar')
plt.xlabel('Countries')
plt.ylabel('Count')
plt.title('countries contribution')
plt.show()
'''#country contribution using scatter
content_counts = df['country'].value_counts().head(10)
plt.figure(figsize=(10,5))
plt.scatter(content_counts.index, content_counts.values)
content_counts.plot(kind='scatter')
plt.xlabel('Countries')
plt.ylabel('Count')
plt.title('countries contribution')
plt.show()'''
#country contribution using lineplot
content_counts = df['country'].value_counts().head(10)
plt.figure(figsize=(10,5))
plt.plot(content_counts.index, content_counts.values)
content_counts.plot(kind='line')
plt.xlabel('Countries')
plt.ylabel('Count')
plt.title('countries contribution')
plt.show()
#Which directors have the most titles on Netflix
director_counts = df['director'].value_counts().head(10)

plt.plot(director_counts.index, director_counts.values)
director_counts.plot(kind='bar')
plt.xlabel('Directors')
plt.ylabel('Count')
plt.title('Directors contribution')
plt.show()
```
