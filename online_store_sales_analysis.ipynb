```python
import numpy as np
import pandas as pd
data={
    "Product_ID":[100,101,102,103,104],
    "Product_Name":["laptop","chair","remote","TV","CPU"],
    "Categories":["electronics","beauty","accesories","electronics","kitchen"],
    "price":[1000,4500,3000,1000,2000],
    "quantity_sold":[10,20,50,30,15]

}
#dataframe
df=pd.DataFrame(data)
print(df)

#handle missing values
df.fillna(0,inplace=True)

#remove duplicates
df.drop_duplicates(inplace=True)

#create revenue
df["Revenue"]=df["price"]*df["quantity_sold"]
print(df)

#products generating revenue above average
avg_revenue=df["Revenue"].mean()
print("\nproducts with revenue above average:")
print(df[df["Revenue"]>avg_revenue])

#sort products by revenue
sorted_df=df.sort_values(by="Revenue",ascending=False)
print("\nSorted Dataset:")
print(sorted_df)

#category wise total revenue
category_revenue=df.groupby("Categories")["Revenue"].sum()
print("\ncategory wise revenue:")
print(category_revenue)

#Numpy statistics on revenue
revenues = np.array(df["Revenue"])
print("\nRevenue Statistics:")
print("Mean:",np.mean(revenues))
print("Median:",np.median(revenues))
print("Standard Deviation:",np.std(revenues))
print("Minimum Revenue:",np.min(revenues))
print("Maximum Revenue:",np.max(revenues))

#best_selling_products
best_selling_products=df[df["quantity_sold"]==df["quantity_sold"].max()]
print("\nBest Selling Products:")
print(best_selling_products)

#sales summary report
print("\nSales Summary Report:")
print("Total Revenue:",df["Revenue"].sum())
print("Total Quantity Sold:",df["quantity_sold"].sum())
print("Average Revenue per Sale:",df["Revenue"].mean())
print("Average Quantity Sold per Sale:",df["quantity_sold"].mean())


```
