```python
#prime number
n = int(input("Enter a number: "))
count = 0
for i in range(1, n+1):
  if(n%i == 0):
    count = count+1
if(count == 2):
  print("prime")
else:
  print("not prime")
```

```python
#palindrome
n = int(input("enter a number: "))
temp = n
reverse = 0
while(n>0):
  reminder = n%10
  reverse = reverse*10 + reminder
  n =n//10
print(reverse)
if(temp == reverse):
  print("palindrome")
else:
  print("not palindrome")
```

```python
#factorial of a number
n = 5
factorial = 1
for i in range(1, n+1):
  factorial = factorial*i
print(factorial)
```

```python
#factorial of n
def factorial(n):
  if n==0 or n==1:
    return 1
  else:
    return n*factorial(n-1)
print(factorial(5))
```

```python
#armstrong number
n = 123
temp = n
l = len(str(n))
sum = 0
while (n>0):
  rem = n%10
  sum = sum+(rem**l)
  n = n//10
print(sum)
if(n == sum):
  print("armstrong")
else:
  print("not armstrong")
```

```python
#LIST
colors = ["red", "blue", "green", "pink", "org", "white"]
colors.append("purple")
print(colors)
colors.insert(2,"burgandy")
print(colors)
colors.remove("white")
print(colors)
colors.pop(2)
print(colors)
l = len(colors)
print(l)
colors.sort()
print(colors)
colors.reverse()
print(colors)
print(colors.count("green"))
```

```python
#SET
names1 = {"ram", "sri", "sita", "priya", "sneha", "tara", "jaya"}
names2 = {"shobha", "radha", "sita", "krishna", "tara", "jaya"}
print(names1)
names2.add("shyam")
print(names2)
names1.remove("sneha")
print(names1)
unions = names1.union(names2)
print(unions)
intersect = names1.intersection(names2)
print(intersect)
diff = names2.difference(names1)
print(diff)
```

```python
#DICTIONARY
person = {
    "name" : "sri",
    "age"  : 21,
    "course" : "IT"
}
print(person)
print(person.keys())
print(person.values())
print(person.items())
print(person.get("age"))
print(person.pop("name"))
print(person)
person.update({"name":"naga"})
print(person)
```
